# EDA - Customer Support Ticket Dataset

## Documentação do dataset

### Fonte

O **Customer Support Ticket Dataset** foi obtido no [Kaggle](https://www.kaggle.com/datasets/suraj520/customer-support-ticket-dataset/data). Neste repositório, o arquivo utilizado está em `data/customer_support_tickets.csv`.

### Principais características

O dataset possui **8.469 chamados e 17 colunas**. Cada linha representa um chamado relacionado a um produto de tecnologia. Os dados incluem informações do cliente e do produto, a descrição do problema, tipo e assunto do chamado, prioridade, canal, status e, quando disponíveis, dados de resposta, resolução e satisfação.

### Motivo da escolha

O dataset foi escolhido por estar alinhado ao problema de atendimento ao cliente e por reunir, no mesmo registro, a descrição do chamado e categorias associadas. Isso permite desenvolver uma classificação de texto sem a necessidade de rotular os dados manualmente. O formato CSV e o tamanho do conjunto também facilitam a análise e a execução local.


## EDA inicial


### Compreensão do problema e do dataset

O projeto busca classificar a intenção dos chamados de suporte. `Ticket Type` e `Ticket Subject` serão tratadas como variáveis-alvo: a primeira representa uma categoria mais ampla do chamado e a segunda, o assunto específico associado ao registro. As demais variáveis do dataset poderão fornecer informações de entrada para apoiar essa classificação.

## Verificação da Qualidade dos Dados

```python
import unicodedata

colunas_categoricas = ["Customer Gender", "Product Purchased", "Ticket Type", "Ticket Subject", "Ticket Status", "Ticket Priority", "Ticket Channel"]

print_formatado("1. Valores ausentes:")
print("Valores ausentes (NaN):")
print(df.isnull().sum())

print("\nValores ausentes representados por '?':")
for coluna in df.select_dtypes(include="object").columns:
    qtd_interrogacao = (df[coluna].str.strip() == "?").sum()
    print(f"{coluna:<30}: {qtd_interrogacao}")
print("Ver Markdown")

print_formatado("2. Registros duplicados:")
print("Duplicatas completas:")
print(df.duplicated().sum())

print("\nDuplicadas desconsiderando Ticket ID:")
print(df.drop(columns=["Ticket ID"]).duplicated().sum())
print("Ver Markdown")

print_formatado("3. Tipos de Dados:")
for coluna in df:
    print(df[coluna].apply(type).value_counts())

print("\nQuantidade de categorias por coluna:")
for coluna in df:
    print(f"{coluna:<30}: {df[coluna].nunique()}")

print("\nDistribuição absoluta:")
for coluna in df:
    if df[coluna].dtype != "object":
        continue
    print(df[coluna].value_counts(), "\n")
print("Ver Markdown")


print_formatado("4. Categorias escritas de maneira diferente:")
def normalizar_texto(valor):
    if pd.isna(valor):
        return valor

    valor = valor.strip().lower()
    valor = (
        unicodedata.normalize('NFKD', valor)
        .encode('ASCII', 'ignore')
        .decode('utf-8')
    )

    return valor

for coluna in colunas_categoricas:
    com_normalizacao = df[coluna].apply(normalizar_texto).nunique()
    sem_normalizacao = df[coluna].nunique()
    print(f"{coluna:<30} -> Com Normalização: {com_normalizacao} | Sem Normalização: {sem_normalizacao}")

print_formatado("5. Textos vazios ou espaços extras:")
for coluna in colunas_categoricas:
    qtd_espacos = (df[coluna] != df[coluna].str.strip()).sum()
    print(f"{coluna:<30}: {qtd_espacos}")

print_formatado("6. Valores impossíveis ou fora do contexto:")
print("\nValores impossíveis:")
print(f"Customer Age mínimo                  : {df['Customer Age'].min()}")
print(f"Customer Age máximo                  : {df['Customer Age'].max()}")
print(f"Customer Age < 18 ou > 100            : {df['Customer Age'][(df['Customer Age'] < 18) | (df['Customer Age'] > 100)].count()}")
print(f"Customer Satisfaction Rating mínimo   : {df['Customer Satisfaction Rating'].min()}")
print(f"Customer Satisfaction Rating máximo   : {df['Customer Satisfaction Rating'].max()}")
print(f"Rating fora do intervalo 1-5          : {df['Customer Satisfaction Rating'][(df['Customer Satisfaction Rating'] < 1) | (df['Customer Satisfaction Rating'] > 5)].count()}")

print("\nDatas fora de um intervalo plausível:")
data_compra = pd.to_datetime(df["Date of Purchase"], errors="coerce")
resposta = pd.to_datetime(df["First Response Time"], errors="coerce")
resolucao = pd.to_datetime(df["Time to Resolution"], errors="coerce")
print(f"Date of Purchase mínima               : {data_compra.min()}")
print(f"Date of Purchase máxima               : {data_compra.max()}")
print(f"Datas de compra não convertidas       : {data_compra.isna().sum() - df['Date of Purchase'].isna().sum()}")

print("\nConsistência temporal entre as colunas de data/hora:")
print(f"First Response Time antes de Date of Purchase  : {(resposta < data_compra).sum()}")
print(f"Time to Resolution antes de First Response Time: {(resolucao < resposta).sum()}")

print("\nInconsistência de capitalização (mesmo valor, cases diferentes):")
for coluna in colunas_categoricas:
    valores_lower = df[coluna].str.lower()
    qtd_original = df[coluna].nunique()
    qtd_lower = valores_lower.nunique()
    if qtd_original != qtd_lower:
        print(f"{coluna:<30}: {qtd_original} categorias originais -> {qtd_lower} após lowercase")
    else:
        print(f"{coluna:<30}: sem inconsistência de capitalização")

print("\nCaracteres inesperados em colunas categóricas (fora de letras, espaço, hífen e '/'):")
for coluna in colunas_categoricas:
    padrao = r"[^A-Za-zÀ-ú\s\-/]"
    qtd_estranhos = df[coluna][df[coluna].str.contains(padrao, na=False, regex=True)].nunique()
    valores_estranhos = df[coluna][df[coluna].str.contains(padrao, na=False, regex=True)].unique()
    if qtd_estranhos > 0:
        print(f"{coluna:<30}: {qtd_estranhos} valor(es) -> {list(valores_estranhos)}")
    else:
        print(f"{coluna:<30}: nenhum caractere inesperado")

print("\nConsistência entre 'Ticket Status' e presença dos campos de resolução:")
fechados_sem_resolucao = df[(df["Ticket Status"] == "Closed") & (df["Resolution"].isna())].shape[0]
abertos_com_resolucao = df[(df["Ticket Status"] != "Closed") & (df["Resolution"].notna())].shape[0]
print(f"Tickets 'Closed' sem Resolution preenchida       : {fechados_sem_resolucao}")
print(f"Tickets não 'Closed' com Resolution preenchida   : {abertos_com_resolucao}")
```

### Análise da Verificação de Qualidade

**1. Valores ausentes**

Não há valores representados por `'?'` — os nulos aparecem só como `NaN` mesmo, em 4 colunas:

- `Resolution`, `Time to Resolution` e `Customer Satisfaction Rating`: 5.700 nulos cada — exatamente os chamados que **ainda não estão `Closed`**. Não é dado faltante por erro de coleta, é ausência estrutural (esses campos só existem quando o ticket é resolvido).

## Hipóteses sobre as Intenções dos Usuários

## Identificação de Outliers e Anomalias

### Variáveis numéricas

```python
colunas_numericas = [
    "Customer Age",
    "Customer Satisfaction Rating",
    "tempo_resolucao_horas"
]

alpha = 0.05
tamanho_maximo_shapiro = 5000

variaveis_normais = []
variaveis_nao_normais = []
resultados_normalidade = {}

print_formatado("1. Selecionar método de identificação de outlier:")

print_formatado("1.1. Q-Q Plot, Assimetria, Shapiro-Wilk")
for coluna in colunas_numericas:

    print(f"\nVariável {coluna}:")
    dados = df[coluna].dropna()
    plt.figure(figsize=(6, 6))

    stats.probplot(
        dados,
        dist="norm",
        plot=plt
    )

    plt.title(f"Q-Q Plot de '{coluna}'")
    plt.xlabel("Quantis teóricos")
    plt.ylabel("Quantis observados")
    plt.tight_layout()
    plt.show()

    assimetria = stats.skew(dados, bias=False, nan_policy="omit")
    print(f"Assimetria             : {assimetria}")

    # ---------------------------------------------------------
    # Teste de Shapiro-Wilk
    # ---------------------------------------------------------
    # Para amostras grandes, são selecionadas no máximo
    # 5.000 observações, pois o teste é muito sensível.
    if len(dados) > tamanho_maximo_shapiro:
        dados_teste = dados.sample(
            n=tamanho_maximo_shapiro,
            random_state=1
        )
    else:
        dados_teste = dados

    estatistica_w, p_valor = stats.shapiro(dados_teste)
    print(f"Teste Shapiro-Wilk, p-valor:{p_valor}")
```

```python
variaveis_irq = [
    "Customer Age",
    "Customer Satisfaction Rating",
    "tempo_resolucao_horas"
]


print_formatado("2. Identificar outliers com IRQ ou Z-Score:")

outliers_por_coluna = {}
indices_outliers = set()

for coluna in variaveis_irq:
    dados = df[coluna].dropna()

    q1 = dados.quantile(0.25)
    q3 = dados.quantile(0.75)
    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    mascara_outliers = ((df[coluna] < limite_inferior) | (df[coluna] > limite_superior))
    outliers_coluna = df.loc[mascara_outliers].copy()
    outliers_por_coluna[coluna] = outliers_coluna
    indices_outliers.update(outliers_coluna.index)

    print(f"\nVariável: {coluna}")
    print(f"Q1: {q1:.2f}")
    print(f"Q3: {q3:.2f}")
    print(f"IQR: {iqr:.2f}")
    print(f"Limite inferior: {limite_inferior:.2f}")
    print(f"Limite superior: {limite_superior:.2f}")
    print(f"Quantidade de outliers: {len(outliers_coluna)}")

    display(outliers_coluna[[coluna]].head(10))
```

#### Observações Identificação de Outliers e Anomalias

    * `Customer Satisfaction Rating`: os pontos formam degraus, pois a variável só assume as notas inteiras de 1 a 5.
    * `tempo_resolucao_horas`: os pontos se afastam da reta nas pontas, com valores concentrados próximos de 0 e cauda à direita.
* `Customer Age` (skew = -0,017) e `Customer Satisfaction Rating` (skew = 0,008) são simétricas, mas isso não indica normalidade, pois a distribuição de ambas é aproximadamente uniforme. `tempo_resolucao_horas` apresentou assimetria positiva moderada (skew = 0,704).
* Nenhuma variável teve p-valor > 0.05, logo nenhuma apresentou indício de distribuição normal.
* Logo, nenhuma variável parece seguir distribuição normal e o método IRQ será utilizado em todas as variáveis.

##### 2. Identificar outliers com IRQ ou ZScore
